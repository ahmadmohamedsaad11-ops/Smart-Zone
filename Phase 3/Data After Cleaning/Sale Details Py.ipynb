Cleaning

```python
df[df["quantity"] <= 0]
```

```python
df = df[df["quantity"] > 0].copy()
```

```python
df[df["unit_cost"] >= df["unit_price"]]
```

```python
products = pd.read_excel(r"D:\Desktop\products PQ.xlsx")
```

```python
products.head()
```

```python
df = df.merge(
    products[["product_id", "unit_price", "unit_cost"]],
    on="product_id",
    how="left",
    suffixes=("_sale", "_product")
)
```

```python
df.loc[
    df["unit_cost_sale"] >= df["unit_price_sale"],
    ["product_id", "unit_price_sale", "unit_price_product",
     "unit_cost_sale", "unit_cost_product"]
]
```

```python
df["unit_price"] = df["unit_price_product"]
df["unit_cost"] = df["unit_cost_product"]
```

```python
df.drop(
    columns=["unit_price_product", "unit_cost_product"],
    inplace=True
)
```

```python
df[df["unit_cost"] >= df["unit_price"]]
```

```python
df[(df["discount"] < 0) | (df["discount"] > 1)]
```

```python
df.loc[
    (df["discount"] < 0) | (df["discount"] > 1),
    "discount"
] = pd.NA
```

```python
df["discount"].isnull().sum()
```

```python
df["discount"] = df["discount"].fillna(df["discount"].mean())
```

```python
df[(df["discount"] < 0) | (df["discount"] > 1)]
```

```python
df.isnull().sum()
```

```python
df.drop(
    columns=["unit_price_sale", "unit_cost_sale"],
    inplace=True
)
```
